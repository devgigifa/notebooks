## 1. Fundamentos

PHP é uma linguagem de script executada no **servidor**. O código fica entre `<?php ... ?>`, instruções terminam com `;` e variáveis começam com `$`.
`echo` imprime texto; `.` concatena strings; comentários: `//`, `#` e `/* */`.

````php
echo "Olá, mundo!\n";

// Comentário de linha
# Outro estilo de comentário
/* Comentário
   de bloco */

$nome = "Ana";
echo "Olá, $nome! Bem-vinda ao PHP.\n";      // aspas duplas interpolam variáveis
echo 'Aspas simples não interpolam: $nome' . "\n";
````

### Variáveis e tipos
PHP é **dinamicamente tipado**, mas aceita tipagem explícita em funções e propriedades.

Tipos escalares: `int`, `float`, `string`, `bool`. Compostos: `array`, `object`. Especiais: `null`.
`var_dump()` mostra tipo e valor; `gettype()` retorna o nome do tipo.

````php
$idade = 20;            // int
$altura = 1.68;         // float
$curso = "Eng. Software"; // string
$ativa = true;          // bool
$nada = null;           // null

var_dump($idade, $altura, $curso, $ativa, $nada);

// Constantes
const PI_APROX = 3.14;
define('ANO', 2026);
echo PI_APROX . " " . ANO . "\n";

// Conversão de tipos (casting)
$texto = "42";
var_dump((int)$texto + 8);   // 50
var_dump((string)3.5);       // "3.5"
var_dump((bool)"");          // false
````

### Operadores
Aritméticos (`+ - * / % **`), comparação (`== === != !== < > <=>`), lógicos (`&& || !`) e **null coalescing** (`??`).

⚠️ Prefira `===` (compara valor **e** tipo) em vez de `==`.

````php
echo 7 + 3, " ", 7 % 3, " ", 2 ** 10, " ", intdiv(7, 2), "\n";

var_dump(0 == "a");      // false no PHP 8
var_dump("1" == "01");   // true  (comparação frouxa)
var_dump("1" === "01");  // false (estrita)
var_dump(1 <=> 2);       // -1 (operador "spaceship")

$apelido = $_GET['apelido'] ?? 'visitante';   // valor padrão se não existir
echo "Olá, $apelido\n";

$x = 5;
$x += 3;  $x *= 2;   // atribuição composta
echo $x, "\n";       // 16
````

## 2. Controle de fluxo

### Condicionais
`if / elseif / else`, `switch` e a expressão **`match`** (PHP 8), que é estrita e retorna valor.

````php
$nota = 7.5;

if ($nota >= 7) {
    echo "Aprovada\n";
} elseif ($nota >= 5) {
    echo "Recuperação\n";
} else {
    echo "Reprovada\n";
}

// Operador ternário
echo $nota >= 7 ? "OK\n" : "Revisar\n";

// match: sem "fallthrough", usa comparação estrita
$dia = 3;
$nomeDia = match ($dia) {
    1 => 'Domingo',
    2 => 'Segunda',
    3 => 'Terça',
    4, 5, 6 => 'Meio da semana',
    default => 'Desconhecido',
};
echo $nomeDia, "\n";
````

### Laços de repetição
`for`, `while`, `do...while` e `foreach` (o mais usado, para arrays). `break` sai do laço e `continue` pula para a próxima iteração.

````php
for ($i = 1; $i <= 5; $i++) {
    echo $i, " ";
}
echo "\n";

$n = 3;
while ($n > 0) {
    echo "Contagem: $n\n";
    $n--;
}

$planetas = ['Mercúrio', 'Vênus', 'Terra', 'Marte'];
foreach ($planetas as $indice => $planeta) {
    if ($planeta === 'Vênus') continue;
    echo "$indice: $planeta\n";
}
````

## 3. Strings e Arrays

### Strings
Funções muito usadas: `strlen`, `strtoupper`, `str_replace`, `substr`, `strpos`, `trim`, `explode`, `implode`, `sprintf`.
Para texto UTF-8 (acentos), prefira as versões `mb_*` (`mb_strlen`, `mb_strtoupper`...).

````php
$frase = "  Programar em PHP é divertido!  ";

echo strlen($frase), " bytes | ", mb_strlen($frase), " caracteres\n";
echo trim($frase), "\n";
echo mb_strtoupper(trim($frase)), "\n";
echo str_replace("PHP", "PHP 8", $frase), "\n";
echo str_contains($frase, "PHP") ? "contém PHP\n" : "não contém\n";

$partes = explode(" ", trim($frase));
echo implode("-", $partes), "\n";

printf("Pi com 2 casas: %.2f | Inteiro com zeros: %05d\n", 3.14159, 42);
echo sprintf("Nome: %s, idade: %d", "Ana", 20), "\n";
````

### Arrays
O array do PHP serve como lista, dicionário e pilha ao mesmo tempo.
- **Indexado**: `[10, 20, 30]`
- **Associativo**: `['chave' => 'valor']`
- **Multidimensional**: arrays dentro de arrays

Funções de ordem superior: `array_map`, `array_filter`, `array_reduce`, além de `usort` para ordenar com regra própria.

````php
$frutas = ['maçã', 'banana'];
$frutas[] = 'uva';                 // adiciona ao final
array_unshift($frutas, 'pera');    // adiciona no início
echo count($frutas), " frutas: ", implode(", ", $frutas), "\n";

$aluno = ['nome' => 'Ana', 'curso' => 'ES', 'notas' => [8, 7.5, 9]];
echo $aluno['nome'], " tem média ", array_sum($aluno['notas']) / count($aluno['notas']), "\n";

$numeros = [1, 2, 3, 4, 5, 6];
$quadrados = array_map(fn($n) => $n ** 2, $numeros);
$pares     = array_filter($numeros, fn($n) => $n % 2 === 0);
$soma      = array_reduce($numeros, fn($acc, $n) => $acc + $n, 0);
print_r($quadrados);
print_r($pares);
echo "Soma: $soma\n";

// Ordenar lista de arrays por campo
$pessoas = [['n' => 'Bia', 'i' => 30], ['n' => 'Caio', 'i' => 22]];
usort($pessoas, fn($a, $b) => $a['i'] <=> $b['i']);
echo $pessoas[0]['n'], "\n";   // Caio

// Spread e desestruturação
[$primeiro, , $terceiro] = [10, 20, 30];
$todos = [...$numeros, 7, 8];
echo "$primeiro $terceiro ", count($todos), "\n";
['nome' => $n, 'curso' => $c] = $aluno;
echo "$n - $c\n";
````

## 4. Funções e escopo

Funções reutilizam código. No PHP moderno use **tipagem** de parâmetros e retorno, **valores padrão**, **argumentos nomeados**, **variádicas** (`...$args`) e **arrow functions** (`fn`).

`declare(strict_types=1);` (no topo de um arquivo) faz o PHP recusar conversões silenciosas de tipo.

````php
function somar(int $a, int $b = 0): int {
    return $a + $b;
}
echo somar(2, 3), " ", somar(5), "\n";

// Argumentos nomeados
function criarUsuario(string $nome, int $idade = 18, bool $ativo = true): string {
    return "$nome ($idade) " . ($ativo ? 'ativo' : 'inativo');
}
echo criarUsuario(nome: "Léo", ativo: false), "\n";

// Variádica
function media(float ...$valores): float {
    return array_sum($valores) / count($valores);
}
echo media(7, 8, 9.5), "\n";

// Tipo anulável e union types
function dobrar(int|float|null $n): int|float {
    return ($n ?? 0) * 2;
}
echo dobrar(null), " ", dobrar(2.5), "\n";

// Funções anônimas e closures
$taxa = 0.1;
$comImposto = fn(float $v) => $v * (1 + $taxa);   // fn captura variáveis externas por valor
echo $comImposto(100), "\n";

$contador = function () {
    static $c = 0;      // variável estática mantém valor entre chamadas
    return ++$c;
};
echo $contador(), $contador(), $contador(), "\n";   // 123
````

### Escopo
Variáveis criadas fora de uma função **não** existem dentro dela (a menos que usadas com `global`, o que deve ser evitado). Passe dados por parâmetro.
Para modificar o original use **referência** (`&`).

````php
$total = 10;

function semAcesso() {
    return isset($total) ? "vê" : "não vê";   // $total não existe aqui
}
echo semAcesso(), "\n";

function incrementar(int &$valor): void {     // por referência
    $valor++;
}
incrementar($total);
echo $total, "\n";   // 11
````

## 5. Orientação a Objetos (POO)

Conceitos-chave:
- **Classe**: molde; **objeto**: instância.
- **Encapsulamento**: visibilidade `public`, `protected`, `private`.
- **Construtor promovido** (PHP 8): declara propriedades direto na assinatura.
- **`readonly`**: propriedade imutável após a criação.
- **Herança** (`extends`), **interfaces** (contrato), **classes abstratas**, **traits** (reuso horizontal), **enums**.

````php
class Produto {
    public function __construct(
        public readonly string $nome,
        private float $preco,
        protected int $estoque = 0,
    ) {}

    public function getPreco(): float { return $this->preco; }

    public function aplicarDesconto(float $percentual): void {
        $this->preco -= $this->preco * ($percentual / 100);
    }

    public function __toString(): string {
        return sprintf("%s - R$ %.2f", $this->nome, $this->preco);
    }
}

$p = new Produto("Teclado", 200.0, 5);
$p->aplicarDesconto(10);
echo $p, "\n";          // Teclado - R$ 180.00
// $p->nome = "x";      // Erro: propriedade readonly
````

### Herança, interfaces, abstração e traits

````php
interface Descontavel {
    public function desconto(): float;
}

trait Loggable {
    public function log(string $msg): void {
        echo "[LOG] " . static::class . ": $msg\n";
    }
}

abstract class Funcionario implements Descontavel {
    use Loggable;

    public function __construct(protected string $nome, protected float $salario) {}

    abstract public function bonus(): float;          // cada filho implementa

    public function total(): float {
        return $this->salario + $this->bonus();
    }
}

class Dev extends Funcionario {
    public function bonus(): float { return $this->salario * 0.2; }
    public function desconto(): float { return $this->salario * 0.11; }
}

class Gerente extends Funcionario {
    public function bonus(): float { return $this->salario * 0.4; }
    public function desconto(): float { return $this->salario * 0.15; }
}

$equipe = [new Dev("Ana", 5000), new Gerente("Bruno", 8000)];
foreach ($equipe as $f) {
    $f->log("total = " . $f->total());          // polimorfismo
}
var_dump($equipe[0] instanceof Descontavel);    // true
````

### Membros estáticos, constantes de classe e enums
`static` pertence à classe (não ao objeto). **Enums** (PHP 8.1) representam um conjunto fixo de valores.

````php
class Contador {
    public const PASSO = 1;
    private static int $total = 0;

    public static function incrementar(): int {
        return self::$total += self::PASSO;
    }
}
Contador::incrementar();
echo Contador::incrementar(), "\n";   // 2

enum Status: string {
    case Pendente = 'pendente';
    case Pago = 'pago';
    case Cancelado = 'cancelado';

    public function rotulo(): string {
        return match ($this) {
            self::Pendente  => '⏳ Aguardando',
            self::Pago      => '✅ Pago',
            self::Cancelado => '❌ Cancelado',
        };
    }
}

$s = Status::from('pago');           // cria a partir do valor
echo $s->rotulo(), " | valor: ", $s->value, "\n";
echo Status::tryFrom('xyz')?->rotulo() ?? 'status inválido', "\n";
````

## 6. Erros e exceções

Use `try / catch / finally` e lance erros com `throw`. Crie exceções próprias estendendo `Exception`.
`Throwable` é a base de tudo (`Exception` e `Error`).

````php
class SaldoInsuficienteException extends Exception {}

class Conta {
    public function __construct(private float $saldo) {}

    public function sacar(float $valor): void {
        if ($valor <= 0) {
            throw new InvalidArgumentException("Valor deve ser positivo");
        }
        if ($valor > $this->saldo) {
            throw new SaldoInsuficienteException("Saldo insuficiente: R$ {$this->saldo}");
        }
        $this->saldo -= $valor;
    }
}

$conta = new Conta(100);
try {
    $conta->sacar(30);
    $conta->sacar(500);
} catch (SaldoInsuficienteException $e) {
    echo "Erro de negócio: ", $e->getMessage(), "\n";
} catch (InvalidArgumentException | TypeError $e) {
    echo "Erro de entrada: ", $e->getMessage(), "\n";
} finally {
    echo "Operação finalizada\n";
}

// Divisão por zero lança DivisionByZeroError (PHP 8)
try {
    echo intdiv(1, 0);
} catch (DivisionByZeroError $e) {
    echo "Capturado: ", $e->getMessage(), "\n";
}
````

## 7. Arquivos e JSON

`file_put_contents` / `file_get_contents` resolvem a maioria dos casos. Para arquivos grandes use `fopen` + `fgets`.
JSON: `json_encode` e `json_decode` (com `JSON_THROW_ON_ERROR` para lançar exceção em vez de falhar em silêncio).

````php
$arquivo = sys_get_temp_dir() . '/tarefas.json';

$tarefas = [
    ['id' => 1, 'titulo' => 'Estudar PHP', 'feito' => false],
    ['id' => 2, 'titulo' => 'Praticar PDO', 'feito' => true],
];

file_put_contents($arquivo, json_encode($tarefas, JSON_PRETTY_PRINT | JSON_UNESCAPED_UNICODE));

if (file_exists($arquivo)) {
    $lidas = json_decode(file_get_contents($arquivo), true, flags: JSON_THROW_ON_ERROR);
    foreach ($lidas as $t) {
        echo ($t['feito'] ? '[x] ' : '[ ] ') . $t['titulo'] . "\n";
    }
}

// CSV linha a linha
$csv = sys_get_temp_dir() . '/dados.csv';
$h = fopen($csv, 'w');
fputcsv($h, ['nome', 'nota']);
fputcsv($h, ['Ana', 9.5]);
fclose($h);

$h = fopen($csv, 'r');
while (($linha = fgetcsv($h)) !== false) {
    echo implode(" | ", $linha), "\n";
}
fclose($h);
````

Para testar localmente: crie um arquivo e rode `php -S localhost:8000` na pasta.
O código abaixo **simula** um formulário para você ver a lógica de validação funcionando aqui.

````php
// Simulando dados que viriam de $_POST
$_POST = ['email' => 'ana@exemplo.com', 'idade' => '20', 'nome' => '<b>Ana</b>'];

$email = filter_var($_POST['email'] ?? '', FILTER_VALIDATE_EMAIL);
$idade = filter_var($_POST['idade'] ?? '', FILTER_VALIDATE_INT, ['options' => ['min_range' => 0, 'max_range' => 120]]);

$erros = [];
if ($email === false) $erros[] = "E-mail inválido";
if ($idade === false) $erros[] = "Idade inválida";

if ($erros) {
    echo implode("\n", $erros), "\n";
} else {
    // Sempre escape a saída em HTML para evitar XSS
    $nomeSeguro = htmlspecialchars($_POST['nome'], ENT_QUOTES, 'UTF-8');
    echo "Cadastro ok: $nomeSeguro ($email, $idade anos)\n";
}
````

## 9. Banco de dados com PDO

**PDO** (PHP Data Objects) é a interface padrão e única para vários bancos (MySQL, PostgreSQL, SQLite...). Trocar de banco muda basicamente a **DSN** (string de conexão).

### Regras de ouro
1. **Sempre** use *prepared statements* (consultas preparadas) → impedem **SQL Injection**.
2. Ative o modo de erro por exceções: `PDO::ERRMODE_EXCEPTION`.
3. Use **transações** quando várias operações precisam funcionar juntas.
4. Nunca coloque senha do banco no código-fonte versionado (use variáveis de ambiente / `.env`).

### 9.1 Conectando
- **SQLite** (arquivo ou memória) — não exige servidor, ideal para estudar.
- **MySQL/MariaDB** — o mais comum em hospedagens PHP.

````php
// ---- SQLite em memória (roda em qualquer lugar com pdo_sqlite) ----
$pdo = new PDO('sqlite::memory:', options: [
    PDO::ATTR_ERRMODE            => PDO::ERRMODE_EXCEPTION,
    PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC,
]);
echo "Conectado ao SQLite! Driver: ", $pdo->getAttribute(PDO::ATTR_DRIVER_NAME), "\n";

/* ---- Como seria no MySQL (exemplo, não executa aqui) ----
$dsn = 'mysql:host=localhost;dbname=meu_banco;charset=utf8mb4';
$pdo = new PDO($dsn, 'usuario', 'senha', [
    PDO::ATTR_ERRMODE            => PDO::ERRMODE_EXCEPTION,
    PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC,
    PDO::ATTR_EMULATE_PREPARES   => false,
]);
*/
````

### 9.2 Criando tabela e inserindo dados (CREATE + INSERT)
`prepare()` monta a consulta com *placeholders* (`?` ou `:nome`) e `execute()` envia os valores separadamente — o banco nunca os interpreta como SQL.

````php
$pdo = new PDO('sqlite::memory:', options: [
    PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION,
    PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC,
]);

$pdo->exec("
    CREATE TABLE usuarios (
        id INTEGER PRIMARY KEY AUTOINCREMENT,   -- no MySQL: INT AUTO_INCREMENT PRIMARY KEY
        nome TEXT NOT NULL,
        email TEXT NOT NULL UNIQUE,
        criado_em TEXT DEFAULT CURRENT_TIMESTAMP
    )
");

$stmt = $pdo->prepare("INSERT INTO usuarios (nome, email) VALUES (:nome, :email)");
$stmt->execute([':nome' => 'Ana',  ':email' => 'ana@exemplo.com']);
$stmt->execute([':nome' => 'Bruno', ':email' => 'bruno@exemplo.com']);

echo "Último ID inserido: ", $pdo->lastInsertId(), "\n";
echo "Linhas afetadas no último insert: ", $stmt->rowCount(), "\n";
````

### 9.3 Consultando (SELECT)
- `fetch()` → uma linha; `fetchAll()` → todas; `fetchColumn()` → um valor.
- Pode iterar com `foreach` direto no *statement* para economizar memória.

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("CREATE TABLE usuarios (id INTEGER PRIMARY KEY AUTOINCREMENT, nome TEXT, email TEXT UNIQUE)");
$ins = $pdo->prepare("INSERT INTO usuarios (nome, email) VALUES (?, ?)");
foreach ([['Ana','ana@x.com'], ['Bruno','bruno@x.com'], ['Carla','carla@x.com']] as $u) $ins->execute($u);

// Todos
foreach ($pdo->query("SELECT * FROM usuarios ORDER BY nome") as $linha) {
    echo "{$linha['id']} - {$linha['nome']} <{$linha['email']}>\n";
}

// Um registro com parâmetro
$busca = $pdo->prepare("SELECT * FROM usuarios WHERE email = :email");
$busca->execute([':email' => 'bruno@x.com']);
$usuario = $busca->fetch();
print_r($usuario);

// Um único valor
echo "Total: ", $pdo->query("SELECT COUNT(*) FROM usuarios")->fetchColumn(), "\n";

// Mapear direto para objetos
class Usuario { public int $id; public string $nome; public string $email; }
$objs = $pdo->query("SELECT * FROM usuarios")->fetchAll(PDO::FETCH_CLASS, Usuario::class);
echo $objs[0]->nome, "\n";
````

### 9.4 Atualizar e excluir (UPDATE / DELETE)
Sempre use `WHERE` (e confira `rowCount()` para saber se algo mudou).

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("CREATE TABLE usuarios (id INTEGER PRIMARY KEY AUTOINCREMENT, nome TEXT, email TEXT)");
$pdo->exec("INSERT INTO usuarios (nome, email) VALUES ('Ana','ana@x.com'), ('Bruno','bruno@x.com')");

$up = $pdo->prepare("UPDATE usuarios SET nome = :nome WHERE id = :id");
$up->execute([':nome' => 'Ana Souza', ':id' => 1]);
echo "Atualizados: ", $up->rowCount(), "\n";

$del = $pdo->prepare("DELETE FROM usuarios WHERE id = ?");
$del->execute([2]);
echo "Removidos: ", $del->rowCount(), "\n";

print_r($pdo->query("SELECT * FROM usuarios")->fetchAll());
````

### 9.5 ⚠️ SQL Injection: o que NÃO fazer
Concatenar entrada do usuário na SQL permite que um atacante altere a consulta.

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("CREATE TABLE usuarios (id INTEGER PRIMARY KEY, nome TEXT, senha TEXT)");
$pdo->exec("INSERT INTO usuarios (nome, senha) VALUES ('admin', 'segredo'), ('ana', '123')");

$entrada = "' OR '1'='1";    // entrada maliciosa

// ❌ VULNERÁVEL
$sql = "SELECT * FROM usuarios WHERE nome = '$entrada'";
echo "Consulta montada: $sql\n";
echo "Linhas retornadas (vulnerável): ", count($pdo->query($sql)->fetchAll()), "\n";   // vaza tudo!

// ✅ SEGURO
$stmt = $pdo->prepare("SELECT * FROM usuarios WHERE nome = ?");
$stmt->execute([$entrada]);
echo "Linhas retornadas (seguro): ", count($stmt->fetchAll()), "\n";                    // 0
````

### 9.6 Transações
Agrupam operações: ou **todas** são confirmadas (`commit`) ou **nenhuma** (`rollBack`). Exemplo clássico: transferência bancária.

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("CREATE TABLE contas (id INTEGER PRIMARY KEY, titular TEXT, saldo REAL)");
$pdo->exec("INSERT INTO contas VALUES (1,'Ana',100), (2,'Bruno',50)");

function transferir(PDO $pdo, int $de, int $para, float $valor): void {
    try {
        $pdo->beginTransaction();

        $saldo = $pdo->prepare("SELECT saldo FROM contas WHERE id = ?");
        $saldo->execute([$de]);
        if ($saldo->fetchColumn() < $valor) {
            throw new RuntimeException("Saldo insuficiente");
        }

        $pdo->prepare("UPDATE contas SET saldo = saldo - ? WHERE id = ?")->execute([$valor, $de]);
        $pdo->prepare("UPDATE contas SET saldo = saldo + ? WHERE id = ?")->execute([$valor, $para]);

        $pdo->commit();
        echo "Transferência de R$ $valor concluída\n";
    } catch (Throwable $e) {
        $pdo->rollBack();
        echo "Falhou e foi revertida: ", $e->getMessage(), "\n";
    }
}

transferir($pdo, 1, 2, 70);    // ok
transferir($pdo, 1, 2, 70);    // falha: saldo insuficiente
print_r($pdo->query("SELECT * FROM contas")->fetchAll());
````

### 9.7 Relacionamentos e JOIN
Tabelas ligadas por chave estrangeira. Aqui: um usuário tem vários posts.

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("PRAGMA foreign_keys = ON");
$pdo->exec("
    CREATE TABLE autores (id INTEGER PRIMARY KEY, nome TEXT NOT NULL);
    CREATE TABLE posts (
        id INTEGER PRIMARY KEY,
        autor_id INTEGER NOT NULL REFERENCES autores(id) ON DELETE CASCADE,
        titulo TEXT NOT NULL
    );
    INSERT INTO autores VALUES (1,'Ana'), (2,'Bruno');
    INSERT INTO posts VALUES (1,1,'Intro ao PHP'), (2,1,'PDO na prática'), (3,2,'Composer');
");

$sql = "
    SELECT a.nome AS autor, COUNT(p.id) AS total, GROUP_CONCAT(p.titulo, ' | ') AS titulos
    FROM autores a
    LEFT JOIN posts p ON p.autor_id = a.id
    GROUP BY a.id
";
foreach ($pdo->query($sql) as $r) {
    echo "{$r['autor']}: {$r['total']} post(s) → {$r['titulos']}\n";
}
````

### 9.8 Mini-projeto: camada de acesso a dados (Repository)
Organizar o SQL numa classe deixa o restante da aplicação limpo e fácil de testar. Este é o padrão usado em projetos reais (e a base para entender frameworks como Laravel).

````php
final class Tarefa {
    public function __construct(
        public readonly ?int $id,
        public readonly string $titulo,
        public readonly bool $concluida = false,
    ) {}
}

final class TarefaRepository {
    public function __construct(private PDO $pdo) {}

    public function criarTabela(): void {
        $this->pdo->exec("CREATE TABLE IF NOT EXISTS tarefas (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            titulo TEXT NOT NULL,
            concluida INTEGER NOT NULL DEFAULT 0
        )");
    }

    public function salvar(Tarefa $t): Tarefa {
        $stmt = $this->pdo->prepare("INSERT INTO tarefas (titulo, concluida) VALUES (?, ?)");
        $stmt->execute([$t->titulo, (int)$t->concluida]);
        return new Tarefa((int)$this->pdo->lastInsertId(), $t->titulo, $t->concluida);
    }

    public function concluir(int $id): bool {
        $stmt = $this->pdo->prepare("UPDATE tarefas SET concluida = 1 WHERE id = ?");
        $stmt->execute([$id]);
        return $stmt->rowCount() > 0;
    }

    /** @return Tarefa[] */
    public function listar(?bool $concluida = null): array {
        $sql = "SELECT * FROM tarefas";
        $params = [];
        if ($concluida !== null) {
            $sql .= " WHERE concluida = ?";
            $params[] = (int)$concluida;
        }
        $stmt = $this->pdo->prepare($sql . " ORDER BY id");
        $stmt->execute($params);
        return array_map(
            fn($r) => new Tarefa((int)$r['id'], $r['titulo'], (bool)$r['concluida']),
            $stmt->fetchAll(PDO::FETCH_ASSOC)
        );
    }

    public function remover(int $id): void {
        $this->pdo->prepare("DELETE FROM tarefas WHERE id = ?")->execute([$id]);
    }
}

$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION]);
$repo = new TarefaRepository($pdo);
$repo->criarTabela();

$a = $repo->salvar(new Tarefa(null, 'Estudar PDO'));
$b = $repo->salvar(new Tarefa(null, 'Fazer o projeto final'));
$repo->concluir($a->id);

echo "Pendentes:\n";
foreach ($repo->listar(false) as $t) echo " - #{$t->id} {$t->titulo}\n";
echo "Concluídas:\n";
foreach ($repo->listar(true) as $t) echo " - #{$t->id} {$t->titulo}\n";
````

## 10. Segurança essencial

| Risco | Defesa |
|---|---|
| **SQL Injection** | Prepared statements (PDO) |
| **XSS** | `htmlspecialchars()` ao imprimir dados em HTML |
| **Senhas expostas** | `password_hash()` / `password_verify()` (nunca MD5/SHA1 puros) |
| **CSRF** | Token em formulários + cookies `SameSite` |
| **Sessão sequestrada** | `session_regenerate_id(true)` após login, HTTPS |
| **Upload malicioso** | Validar tipo/tamanho, renomear, salvar fora da pasta pública |

````php
// Hash seguro de senha
$hash = password_hash("minhaSenha123", PASSWORD_DEFAULT);
echo "Hash: ", substr($hash, 0, 20), "...\n";

var_dump(password_verify("minhaSenha123", $hash));   // true
var_dump(password_verify("errada", $hash));          // false

// Token aleatório (CSRF, reset de senha etc.)
$token = bin2hex(random_bytes(16));
echo "Token: $token\n";

// Comparação segura contra timing attacks
var_dump(hash_equals($token, $token));
````

### Mini-sistema de login com banco (juntando tudo)

````php
$pdo = new PDO('sqlite::memory:', options: [PDO::ATTR_ERRMODE => PDO::ERRMODE_EXCEPTION, PDO::ATTR_DEFAULT_FETCH_MODE => PDO::FETCH_ASSOC]);
$pdo->exec("CREATE TABLE usuarios (id INTEGER PRIMARY KEY, email TEXT UNIQUE NOT NULL, senha_hash TEXT NOT NULL)");

function registrar(PDO $pdo, string $email, string $senha): bool {
    if (!filter_var($email, FILTER_VALIDATE_EMAIL) || strlen($senha) < 8) return false;
    try {
        $pdo->prepare("INSERT INTO usuarios (email, senha_hash) VALUES (?, ?)")
            ->execute([$email, password_hash($senha, PASSWORD_DEFAULT)]);
        return true;
    } catch (PDOException $e) {
        return false;   // e-mail duplicado, por exemplo
    }
}

function autenticar(PDO $pdo, string $email, string $senha): bool {
    $stmt = $pdo->prepare("SELECT senha_hash FROM usuarios WHERE email = ?");
    $stmt->execute([$email]);
    $hash = $stmt->fetchColumn();
    return $hash !== false && password_verify($senha, $hash);
}

var_dump(registrar($pdo, "ana@exemplo.com", "senhaForte123"));   // true
var_dump(registrar($pdo, "ana@exemplo.com", "outraSenha123"));   // false (duplicado)
var_dump(autenticar($pdo, "ana@exemplo.com", "senhaForte123"));  // true
var_dump(autenticar($pdo, "ana@exemplo.com", "errada"));         // false
````

## 11. Namespaces, autoload e Composer

Em projetos reais o código é dividido em arquivos/classes. **Namespaces** evitam conflito de nomes e o **Composer** (gerenciador de dependências) faz o *autoload* seguindo o padrão **PSR-4**.

**Estrutura típica**
```
meu-projeto/
├── composer.json
├── public/index.php        ← ponto de entrada
├── src/
│   ├── Model/Usuario.php   ← namespace App\Model;
│   └── Repository/UsuarioRepository.php
└── vendor/                 ← gerado pelo Composer
```

**composer.json**
```json
{
  "autoload": { "psr-4": { "App\\": "src/" } },
  "require": { "vlucas/phpdotenv": "^5.6" }
}
```

**Comandos**
```bash
composer init
composer require vlucas/phpdotenv   # instala pacote
composer dump-autoload              # regenera o autoload
```

**src/Model/Usuario.php**
```php
<?php
declare(strict_types=1);
namespace App\Model;

final class Usuario {
    public function __construct(public string $nome) {}
}
```

**public/index.php**
```php
<?php
require __DIR__ . '/../vendor/autoload.php';

use App\Model\Usuario;

$u = new Usuario("Ana");
echo $u->nome;
```

## 12. Recursos modernos que vale conhecer

Pequena vitrine de recursos do PHP 8.x.

````php
// Nullsafe operator
class Endereco { public ?string $cidade = "Florianópolis"; }
class Pessoa { public ?Endereco $endereco = null; }
$p = new Pessoa();
echo $p->endereco?->cidade ?? "sem endereço", "\n";

// Generators: processam grandes volumes sem estourar memória
function numerosGrandes(int $max) {
    for ($i = 1; $i <= $max; $i++) yield $i;
}
$soma = 0;
foreach (numerosGrandes(1_000_000) as $n) $soma += $n;
echo "Soma até 1 milhão: $soma\n";

// Funções de array úteis
$dados = [['n'=>'Ana','c'=>'ES'], ['n'=>'Bia','c'=>'SI'], ['n'=>'Caio','c'=>'ES']];
print_r(array_column($dados, 'n', 'n'));
$porCurso = [];
foreach ($dados as $d) $porCurso[$d['c']][] = $d['n'];
print_r($porCurso);

// Datas
$hoje = new DateTimeImmutable('2026-11-04', new DateTimeZone('America/Sao_Paulo'));
echo $hoje->modify('+30 days')->format('d/m/Y'), "\n";
````
