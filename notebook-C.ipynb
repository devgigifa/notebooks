## 📌 1 — Pensamento Computacional & Primeiro Programa <a name='s1'></a>

**Teoria:**  
Todo programa C começa pela função `main()`. O compilador transforma o código-fonte (`.c`) em um executável binário.  
Fluxo: **Escrever → Pré-processar → Compilar → Linkar → Executar**

```
┌──────────────┐    gcc     ┌──────────────┐    ./    ┌─────────┐
│   hello.c    │ ─────────▶ │   hello      │ ───────▶ │ Saída   │
│  (código C)  │            │  (binário)   │          │ terminal│
└──────────────┘            └──────────────┘          └─────────┘
```

````python
%%shell
cat << 'EOF' > hello.c
#include <stdio.h>

/* main() é o ponto de entrada de todo programa C */
int main(void) {
    printf("Olá, Mundo!\n");
    printf("Bem-vindo ao Guia de C!\n");
    return 0;  /* 0 = execução bem-sucedida */
}
EOF
gcc hello.c -o hello && ./hello
````

## 📌 2 — Tipos de Dados <a name='s2'></a>

**Teoria:**  
C é estaticamente tipada: o tipo define quantos bytes são reservados na memória.

| Tipo | Bytes | Faixa típica | Uso |
|------|-------|-------------|-----|
| `char` | 1 | -128 a 127 | Caracteres |
| `int` | 4 | -2.147.483.648 a 2.147.483.647 | Inteiros |
| `float` | 4 | ±3.4 × 10³⁸ | Ponto flutuante simples |
| `double` | 8 | ±1.8 × 10³⁰⁸ | Ponto flutuante duplo |

⚠️ **Divisão inteira**: `7 / 2 = 3` (não 3.5). Use cast para `double` quando precisar de fração.

````python
%%shell
cat << 'EOF' > tipos.c
#include <stdio.h>

int main(void) {
    int    idade   = 25;
    float  preco   = 19.90f;      /* sufixo 'f' para float literal */
    double pi      = 3.14159265;
    char   letra   = 'A';
    const  int ANO = 2026;        /* constante — não pode ser alterada */

    printf("Idade:  %d\n",   idade);
    printf("Preco:  %.2f\n", preco);
    printf("Pi:     %.8f\n", pi);
    printf("Letra:  %c  (ASCII %d)\n", letra, letra);
    printf("Ano:    %d\n",   ANO);

    /* Tamanhos em bytes */
    printf("\nTamanhos:\n");
    printf("  char:   %zu bytes\n", sizeof(char));
    printf("  int:    %zu bytes\n", sizeof(int));
    printf("  float:  %zu bytes\n", sizeof(float));
    printf("  double: %zu bytes\n", sizeof(double));

    /* Casting explícito */
    int a = 7, b = 2;
    printf("\n7 / 2 (int)    = %d\n", a / b);              /* 3 */
    printf("7 / 2 (double) = %.1f\n", (double)a / b);      /* 3.5 */

    return 0;
}
EOF
gcc tipos.c -o tipos && ./tipos
````

## 📌 3 — Operadores <a name='s3'></a>

**Teoria:**  
- **Aritméticos**: `+ - * / %` — `%` é o resto da divisão inteira  
- **Relacionais**: `== != < > <= >=` — retornam `1` (verdadeiro) ou `0` (falso)  
- **Lógicos**: `&&` (E), `||` (OU), `!` (NÃO)  
- **Atribuição composta**: `+= -= *= /= %=`  

**Tabela Verdade:**

| A | B | A && B | A \|\| B | !A |
|---|---|--------|----------|----|
| 0 | 0 | 0 | 0 | 1 |
| 0 | 1 | 0 | 1 | 1 |
| 1 | 0 | 0 | 1 | 0 |
| 1 | 1 | 1 | 1 | 0 |

````python
%%shell
cat << 'EOF' > operadores.c
#include <stdio.h>

int main(void) {
    int x = 10, y = 3;

    printf("=== Aritméticos ===\n");
    printf("%d + %d = %d\n", x, y, x + y);   /* 13 */
    printf("%d - %d = %d\n", x, y, x - y);   /* 7  */
    printf("%d * %d = %d\n", x, y, x * y);   /* 30 */
    printf("%d / %d = %d\n", x, y, x / y);   /* 3  (inteiro!) */
    printf("%d %% %d = %d\n", x, y, x % y);  /* 1  (resto) */

    printf("\n=== Relacionais (1=true, 0=false) ===\n");
    printf("10 == 3 : %d\n", x == y);   /* 0 */
    printf("10 != 3 : %d\n", x != y);   /* 1 */
    printf("10 >  3 : %d\n", x >  y);   /* 1 */

    printf("\n=== Lógicos ===\n");
    int chuva = 1, guarda_chuva = 0;
    printf("chuva && !guarda_chuva = %d\n", chuva && !guarda_chuva); /* 1 */
    printf("!chuva || guarda_chuva = %d\n", !chuva || guarda_chuva); /* 0 */

    printf("\n=== Atribuição composta ===\n");
    x += 5;  printf("x += 5  → x = %d\n", x); /* 15 */
    x *= 2;  printf("x *= 2  → x = %d\n", x); /* 30 */
    x %= 7;  printf("x %%= 7  → x = %d\n", x); /* 2  */

    printf("\n=== Pré vs Pós-incremento ===\n");
    int i = 5;
    printf("i = %d\n", i);
    printf("i++ usa %d, depois i = %d\n", i++, i);
    printf("++i retorna %d\n", ++i);

    return 0;
}
EOF
gcc operadores.c -o operadores && ./operadores
````

## 📌 4 — Entrada e Saída <a name='s4'></a>

**Teoria:**  
- `printf(formato, ...)` — exibe dados formatados  
- `scanf(formato, &var)` — lê dados do usuário (**sempre use `&`** para variáveis simples, exceto strings)  
- `fgets(str, tam, stdin)` — lê linha inteira com espaços

| Especificador | Tipo | Exemplo |
|---|---|---|
| `%d` | `int` | `42` |
| `%.2f` | `float/double` | `3.14` |
| `%c` | `char` | `A` |
| `%s` | string | `hello` |
| `%zu` | `size_t` | `8` |

> 💡 Como o Colab não tem terminal interativo, simulamos a entrada com `echo` redirecionando para `stdin`.

````python
%%shell
cat << 'EOF' > io.c
#include <stdio.h>
#include <string.h>

int main(void) {
    int   idade;
    float altura;
    char  nome[50];

    /* No Colab a entrada vem do pipe abaixo */
    printf("Digite seu nome: ");
    scanf("%49s", nome);

    printf("Digite sua idade: ");
    scanf("%d", &idade);

    printf("Digite sua altura (m): ");
    scanf("%f", &altura);

    printf("\n--- Resultado ---\n");
    printf("Nome:   %s\n",   nome);
    printf("Idade:  %d anos\n", idade);
    printf("Altura: %.2f m\n",  altura);

    /* Formatação avançada */
    printf("\n--- Formatação ---\n");
    printf("|%-20s|%6.2f|\n", "Item A",  12.5);   /* alinhado à esquerda */
    printf("|%-20s|%6.2f|\n", "Item B", 999.99);   /* alinhado à direita  */
    printf("|%020d|\n", 42);                        /* zeros à esquerda    */

    return 0;
}
EOF
gcc io.c -o io
# Simula entrada do usuário via pipe
echo -e "Maria\n22\n1.68" | ./io
````

## 📌 5 — Estruturas de Decisão <a name='s5'></a>

**Teoria:**  
- `if / else if / else` — condições booleanas (qualquer valor != 0 é verdadeiro)  
- `switch / case` — comparação de valor inteiro com múltiplas alternativas fixas  
⚠️ Sempre use `break` no `switch`, ou o código cai no próximo case (*fall-through*).

````python
%%shell
cat << 'EOF' > decisao.c
#include <stdio.h>

int main(void) {
    /* ── if-else encadeado: classificação de notas ── */
    int notas[] = {95, 82, 68, 51, 30};
    int n = 5;

    printf("=== Classificação de Notas ===\n");
    for (int i = 0; i < n; i++) {
        int nota = notas[i];
        char *conceito;

        if      (nota >= 90) conceito = "A — Excelente";
        else if (nota >= 70) conceito = "B — Bom";
        else if (nota >= 50) conceito = "C — Regular";
        else                 conceito = "D — Reprovado";

        printf("Nota %3d → %s\n", nota, conceito);
    }

    /* ── switch-case: dias da semana ── */
    printf("\n=== Dias da Semana ===\n");
    for (int dia = 1; dia <= 7; dia++) {
        switch (dia) {
            case 1: printf("Dia %d: Segunda-feira\n", dia); break;
            case 2: printf("Dia %d: Terça-feira\n",   dia); break;
            case 3: printf("Dia %d: Quarta-feira\n",  dia); break;
            case 4: printf("Dia %d: Quinta-feira\n",  dia); break;
            case 5: printf("Dia %d: Sexta-feira\n",   dia); break;
            case 6: /* fall-through intencional */
            case 7: printf("Dia %d: Fim de semana!\n", dia); break;
            default: printf("Dia inválido\n");
        }
    }

    /* ── Operador ternário ── */
    int x = 42;
    printf("\n%d é %s\n", x, (x % 2 == 0) ? "par" : "ímpar");

    return 0;
}
EOF
gcc decisao.c -o decisao && ./decisao
````

## 📌 6 — Estruturas de Repetição <a name='s6'></a>

**Teoria:**

| Estrutura | Quando usar | Testa condição |
|-----------|------------|----------------|
| `for` | Nº de iterações conhecido | Antes |
| `while` | Nº desconhecido, pode não executar | Antes |
| `do-while` | Deve executar pelo menos 1 vez | Depois |

- `break` — sai imediatamente do laço  
- `continue` — pula para a próxima iteração

````python
%%shell
cat << 'EOF' > loops.c
#include <stdio.h>

int main(void) {
    /* ── for: tabuada do 7 ── */
    printf("=== Tabuada do 7 ===\n");
    for (int i = 1; i <= 10; i++)
        printf("7 x %2d = %2d\n", i, 7 * i);

    /* ── for: triângulo de asteriscos ── */
    printf("\n=== Triângulo ===\n");
    for (int i = 1; i <= 5; i++) {
        for (int j = 0; j < i; j++) printf("* ");
        printf("\n");
    }

    /* ── while: sequência de Fibonacci ── */
    printf("\n=== Fibonacci (até 100) ===\n");
    int a = 0, b = 1;
    while (a <= 100) {
        printf("%d ", a);
        int t = a + b;
        a = b;
        b = t;
    }
    printf("\n");

    /* ── for + continue: números ímpares ── */
    printf("\n=== Ímpares de 1 a 20 ===\n");
    for (int i = 1; i <= 20; i++) {
        if (i % 2 == 0) continue;  /* pula pares */
        printf("%d ", i);
    }
    printf("\n");

    /* ── for + break: primeiro múltiplo de 7 acima de 50 ── */
    printf("\n=== Primeiro múltiplo de 7 > 50 ===\n");
    for (int i = 51; i < 200; i++) {
        if (i % 7 == 0) {
            printf("%d\n", i);
            break;
        }
    }

    /* ── do-while: menu simulado ── */
    printf("\n=== Validação do-while ===\n");
    int tentativas[] = {5, 0, -1, 2};  /* simula entradas inválidas depois válida */
    int idx = 0, opcao;
    do {
        opcao = tentativas[idx++];
        printf("Tentativa: %d %s\n", opcao, (opcao == 1 || opcao == 2) ? "✓" : "✗");
    } while (opcao != 1 && opcao != 2 && idx < 4);

    return 0;
}
EOF
gcc loops.c -o loops && ./loops
````

## 📌 7 — Funções <a name='s7'></a>

**Teoria:**  
- **Protótipo**: declaração antecipada que informa o compilador sobre o tipo de retorno e parâmetros  
- **Passagem por valor**: a função recebe uma *cópia* — alterar dentro não altera o original  
- **Passagem por ponteiro**: a função recebe o endereço — altera o original  
- **Escopo local**: variáveis declaradas dentro de uma função existem somente ali  
- **Escopo global**: variáveis declaradas fora de qualquer função são visíveis em todo o arquivo

````python
%%shell
cat << 'EOF' > funcoes.c
#include <stdio.h>

/* ── Protótipos ── */
int    fatorial(int n);
double media(double v[], int tam);
void   trocar(int *a, int *b);    /* passagem por ponteiro */
int    eh_primo(int n);

/* variável global */
int chamadas = 0;

int main(void) {
    /* Fatorial */
    printf("=== Fatorial ===\n");
    for (int i = 0; i <= 10; i++)
        printf("%2d! = %d\n", i, fatorial(i));

    /* Média */
    double notas[] = {7.5, 8.0, 9.5, 6.0, 8.5};
    printf("\nMédia: %.2f\n", media(notas, 5));

    /* Trocar — passagem por referência */
    int x = 10, y = 20;
    printf("\nAntes: x=%d y=%d\n", x, y);
    trocar(&x, &y);
    printf("Depois: x=%d y=%d\n", x, y);

    /* Primos */
    printf("\nPrimos de 2 a 50: ");
    for (int i = 2; i <= 50; i++)
        if (eh_primo(i)) printf("%d ", i);
    printf("\n");

    printf("\nFunções chamadas: %d\n", chamadas);
    return 0;
}

/* ── Definições ── */
int fatorial(int n) {
    chamadas++;
    if (n <= 1) return 1;
    return n * fatorial(n - 1);  /* recursão */
}

double media(double v[], int tam) {
    chamadas++;
    double s = 0;
    for (int i = 0; i < tam; i++) s += v[i];
    return s / tam;
}

void trocar(int *a, int *b) {
    chamadas++;
    int tmp = *a;
    *a = *b;
    *b = tmp;
}

int eh_primo(int n) {
    chamadas++;
    if (n < 2) return 0;
    for (int i = 2; i * i <= n; i++)
        if (n % i == 0) return 0;
    return 1;
}
EOF
gcc funcoes.c -o funcoes && ./funcoes
````

## 📌 8 — Vetores e Matrizes <a name='s8'></a>

**Teoria:**  
- **Vetor** (array 1D): posições contíguas de memória, indexadas a partir de `0`  
- **Matriz** (array 2D): vetor de vetores — `mat[linha][coluna]`  
- Tamanho do array: `sizeof(arr) / sizeof(arr[0])`  
⚠️ C não verifica limites — acessar `v[n]` em array de `n` posições é *undefined behavior*.

````python
%%shell
cat << 'EOF' > arrays.c
#include <stdio.h>

/* Ordena com Bubble Sort (didático) */
void bubble_sort(int v[], int n) {
    for (int i = 0; i < n - 1; i++)
        for (int j = 0; j < n - 1 - i; j++)
            if (v[j] > v[j+1]) {
                int t = v[j]; v[j] = v[j+1]; v[j+1] = t;
            }
}

int main(void) {
    /* ── Vetor ── */
    int notas[] = {85, 42, 97, 61, 73, 55, 88};
    int n = sizeof(notas) / sizeof(notas[0]);

    printf("=== Vetor original ===\n");
    for (int i = 0; i < n; i++) printf("%d ", notas[i]);
    printf("\n");

    bubble_sort(notas, n);
    printf("=== Após Bubble Sort ===\n");
    for (int i = 0; i < n; i++) printf("%d ", notas[i]);
    printf("\n");

    /* ── Matriz identidade 4x4 ── */
    printf("\n=== Matriz Identidade 4x4 ===\n");
    int mat[4][4];
    for (int i = 0; i < 4; i++)
        for (int j = 0; j < 4; j++)
            mat[i][j] = (i == j) ? 1 : 0;

    for (int i = 0; i < 4; i++) {
        for (int j = 0; j < 4; j++) printf("%2d", mat[i][j]);
        printf("\n");
    }

    /* ── Soma de duas matrizes 2x2 ── */
    printf("\n=== Soma de Matrizes ===\n");
    int A[2][2] = {{1,2},{3,4}};
    int B[2][2] = {{5,6},{7,8}};
    int C[2][2];
    for (int i = 0; i < 2; i++)
        for (int j = 0; j < 2; j++)
            C[i][j] = A[i][j] + B[i][j];

    for (int i = 0; i < 2; i++) {
        printf("|%2d %2d|  +  |%2d %2d|  =  |%2d %2d|\n",
               A[i][0],A[i][1], B[i][0],B[i][1], C[i][0],C[i][1]);
    }

    return 0;
}
EOF
gcc arrays.c -o arrays && ./arrays
````

## 📌 9 — Ponteiros <a name='s9'></a>

**Teoria:**  
- `&x` → endereço de memória de `x`  
- `int *p = &x` → `p` armazena o endereço de `x`  
- `*p` → acessa/modifica o valor no endereço (desreferência)  
- O nome de um array já é um ponteiro para o primeiro elemento: `v == &v[0]`

```
   x = 42          p
 ┌────────┐      ┌────────┐
 │   42   │◀──── │ 0x1004 │
 └────────┘      └────────┘
  endereço        p guarda
  0x1004          o endereço
```

````python
%%shell
cat << 'EOF' > ponteiros.c
#include <stdio.h>

void dobrar(int *p)   { *p *= 2; }
void zerar(int *v, int n) {
    for (int i = 0; i < n; i++) v[i] = 0;
}

int main(void) {
    /* Fundamentos */
    int x = 42;
    int *p = &x;

    printf("=== Fundamentos ===\n");
    printf("x     = %d\n", x);
    printf("&x    = %p\n", (void*)&x);
    printf("p     = %p  (mesmo endereço)\n", (void*)p);
    printf("*p    = %d  (valor via ponteiro)\n", *p);

    *p = 99;
    printf("\nApós *p = 99:\n");
    printf("x = %d  (alterado via ponteiro!)\n", x);

    dobrar(&x);
    printf("Após dobrar(&x): x = %d\n", x);

    /* Aritmética de ponteiros */
    int v[] = {10, 20, 30, 40, 50};
    int *pv = v;   /* aponta para v[0] */

    printf("\n=== Aritmética de Ponteiros ===\n");
    for (int i = 0; i < 5; i++)
        printf("pv+%d → endereço %p → valor %d\n",
               i, (void*)(pv+i), *(pv+i));

    /* Ponteiro para ponteiro */
    int **pp = &p;
    printf("\n=== Ponteiro para Ponteiro ===\n");
    printf("**pp = %d\n", **pp);

    /* Zerar array via ponteiro */
    zerar(v, 5);
    printf("\n=== Após zerar(v, 5) ===\n");
    for (int i = 0; i < 5; i++) printf("%d ", v[i]);
    printf("\n");

    return 0;
}
EOF
gcc ponteiros.c -o ponteiros && ./ponteiros
````

## 📌 10 — Strings <a name='s10'></a>

**Teoria:**  
Strings em C são `char[]` terminados com `'\0'`. A biblioteca `string.h` oferece:

| Função | O que faz |
|--------|----------|
| `strlen(s)` | Comprimento (sem `'\0'`) |
| `strcpy(dst, src)` | Copia `src` em `dst` |
| `strncpy(dst, src, n)` | Copia no máximo `n` bytes (mais seguro) |
| `strcat(dst, src)` | Concatena `src` ao final de `dst` |
| `strcmp(a, b)` | Compara: 0 = iguais, <0 = a<b, >0 = a>b |
| `strchr(s, c)` | Ponteiro para primeira ocorrência de `c` em `s` |
| `strstr(s, sub)` | Ponteiro para primeira ocorrência de `sub` em `s` |

````python
%%shell
cat << 'EOF' > strings.c
#include <stdio.h>
#include <string.h>
#include <ctype.h>   /* toupper, tolower, isdigit... */

/* Converte string para maiúsculas (in-place) */
void para_maiuscula(char *s) {
    for (int i = 0; s[i]; i++) s[i] = toupper(s[i]);
}

/* Conta vogais */
int contar_vogais(const char *s) {
    int count = 0;
    for (int i = 0; s[i]; i++)
        if (strchr("aeiouAEIOU", s[i])) count++;
    return count;
}

int main(void) {
    char s1[80] = "Programação";
    char s2[80] = " em C";
    char s3[80];

    printf("=== Funções Básicas ===\n");
    printf("strlen(\"%s\") = %zu\n", s1, strlen(s1));

    strcpy(s3, s1);
    strcat(s3, s2);
    printf("strcpy + strcat: \"%s\"\n", s3);

    printf("\n=== strcmp ===\n");
    char *palavras[] = {"banana", "abacaxi", "banana", "cereja"};
    for (int i = 0; i < 3; i++) {
        int cmp = strcmp(palavras[i], palavras[i+1]);
        printf("\"%s\" vs \"%s\": %d → %s\n",
               palavras[i], palavras[i+1], cmp,
               cmp < 0 ? "menor" : cmp > 0 ? "maior" : "igual");
    }

    printf("\n=== Manipulação ===\n");
    char frase[] = "hello world";
    printf("Original: %s\n", frase);
    para_maiuscula(frase);
    printf("Maiúscula: %s\n", frase);
    printf("Vogais em 'Programacao': %d\n", contar_vogais("Programacao"));

    /* strstr */
    char texto[] = "O rato roeu a roupa do rei";
    char *pos = strstr(texto, "roupa");
    if (pos) printf("\n'roupa' encontrada na posição %ld\n", pos - texto);

    return 0;
}
EOF
gcc strings.c -o strings && ./strings
````

## 📌 11 — Alocação Dinâmica de Memória <a name='s11'></a>

**Teoria:**

| Área | Característica |
|------|---------------|
| **Stack** (pilha) | Variáveis locais, tamanho fixo em compilação, liberada ao sair da função |
| **Heap** | `malloc`/`calloc`, tamanho definido em runtime, deve ser liberada com `free` |

- `malloc(n)` → aloca `n` bytes, conteúdo indefinido  
- `calloc(qtd, size)` → aloca e **zera** a memória  
- `realloc(p, novo_n)` → redimensiona bloco existente  
- `free(p)` → libera o bloco  
⚠️ Sempre verifique se `malloc` retornou `NULL`. Sempre faça `free`. Use Valgrind para detectar vazamentos.

````python
%%shell
cat << 'EOF' > dinamico.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

int comparar(const void *a, const void *b) {
    return (*(int*)a - *(int*)b);
}

int main(void) {
    int n = 8;

    /* ── malloc ── */
    int *v = (int*) malloc(n * sizeof(int));
    if (!v) { fprintf(stderr, "Sem memória\n"); return 1; }

    /* Preenche com valores */
    int valores[] = {64, 25, 12, 22, 11, 90, 45, 3};
    memcpy(v, valores, n * sizeof(int));

    printf("=== malloc: array alocado no heap ===\n");
    for (int i = 0; i < n; i++) printf("%d ", v[i]);
    printf("\n");

    /* Ordena com qsort (stdlib) */
    qsort(v, n, sizeof(int), comparar);
    printf("=== Após qsort ===\n");
    for (int i = 0; i < n; i++) printf("%d ", v[i]);
    printf("\n");

    /* ── realloc: dobra o tamanho ── */
    int novo_n = n * 2;
    v = (int*) realloc(v, novo_n * sizeof(int));
    if (!v) { fprintf(stderr, "Sem memória no realloc\n"); return 1; }
    for (int i = n; i < novo_n; i++) v[i] = i * 10;

    printf("\n=== Após realloc (tamanho dobrado) ===\n");
    for (int i = 0; i < novo_n; i++) printf("%d ", v[i]);
    printf("\n");

    free(v);
    v = NULL;   /* evita dangling pointer */

    /* ── calloc: inicializa com zeros ── */
    printf("\n=== calloc: 5 doubles zerados ===\n");
    double *notas = (double*) calloc(5, sizeof(double));
    for (int i = 0; i < 5; i++) printf("%.1f ", notas[i]);  /* todos 0.0 */
    printf("\n");
    free(notas);

    printf("\nMemória liberada com sucesso!\n");
    return 0;
}
EOF
gcc dinamico.c -o dinamico && ./dinamico
````

## 📌 12 — Structs <a name='s12'></a>

**Teoria:**  
- `struct` agrupa variáveis de tipos diferentes em um tipo composto  
- `typedef struct { ... } Nome;` cria um alias conveniente  
- Membro de variável: `aluno.nome` (operador ponto)  
- Membro via ponteiro: `p->nome` (operador seta) equivale a `(*p).nome`

````python
%%shell
cat << 'EOF' > structs.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

typedef struct {
    int    matricula;
    char   nome[60];
    double notas[3];
    double media;  /* calculada */
} Aluno;

/* Lista ligada simples */
typedef struct No {
    int        valor;
    struct No *prox;
} No;

Aluno criar_aluno(int mat, const char *nome, double n1, double n2, double n3) {
    Aluno a;
    a.matricula = mat;
    strncpy(a.nome, nome, 59);
    a.notas[0] = n1; a.notas[1] = n2; a.notas[2] = n3;
    a.media = (n1 + n2 + n3) / 3.0;
    return a;
}

void exibir(const Aluno *a) {
    printf("  [%04d] %-20s  notas: %.1f %.1f %.1f  média: %.2f  %s\n",
           a->matricula, a->nome,
           a->notas[0], a->notas[1], a->notas[2],
           a->media,
           a->media >= 7.0 ? "✓ Aprovado" : "✗ Reprovado");
}

int main(void) {
    Aluno turma[] = {
        criar_aluno(1001, "Ana Souza",    8.0, 9.0, 8.5),
        criar_aluno(1002, "Bruno Lima",   6.0, 7.0, 5.5),
        criar_aluno(1003, "Carla Neves",  9.5, 9.0, 10.0),
        criar_aluno(1004, "Diego Melo",   4.0, 5.0, 6.0),
    };
    int n = 4;

    printf("=== Boletim da Turma ===\n");
    for (int i = 0; i < n; i++) exibir(&turma[i]);

    /* Melhor aluno */
    Aluno *melhor = &turma[0];
    for (int i = 1; i < n; i++)
        if (turma[i].media > melhor->media) melhor = &turma[i];
    printf("\nMelhor aluno: %s (%.2f)\n", melhor->nome, melhor->media);

    /* Lista ligada simples */
    printf("\n=== Lista Ligada ===\n");
    No *head = NULL;
    for (int v = 1; v <= 5; v++) {
        No *novo = (No*) malloc(sizeof(No));
        novo->valor = v * 10;
        novo->prox  = head;
        head = novo;
    }
    for (No *cur = head; cur; cur = cur->prox)
        printf("%d → ", cur->valor);
    printf("NULL\n");

    /* Libera lista */
    while (head) { No *t = head->prox; free(head); head = t; }

    return 0;
}
EOF
gcc structs.c -o structs && ./structs
````

## 📌 13 — Manipulação de Arquivos <a name='s13'></a>

**Teoria:**

| Modo | Significado |
|------|-------------|
| `"r"` | Leitura; falha se não existir |
| `"w"` | Escrita; cria ou trunca |
| `"a"` | Append; adiciona ao final |
| `"r+"` | Leitura e escrita |
| `"rb"/"wb"` | Binário |

Fluxo: `fopen` → operações → `fclose`. Sempre verifique se `fopen` retornou `NULL`.

````python
%%shell
cat << 'EOF' > arquivos.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>

typedef struct { char nome[50]; double media; } Aluno;

void salvar_csv(const char *path, Aluno *v, int n) {
    FILE *f = fopen(path, "w");
    if (!f) { perror("fopen"); exit(1); }
    fprintf(f, "nome,media\n");
    for (int i = 0; i < n; i++)
        fprintf(f, "%s,%.2f\n", v[i].nome, v[i].media);
    fclose(f);
    printf("Salvo: %s\n", path);
}

void ler_csv(const char *path) {
    FILE *f = fopen(path, "r");
    if (!f) { perror("fopen"); exit(1); }
    char linha[100];
    printf("\nConteúdo de %s:\n", path);
    while (fgets(linha, sizeof(linha), f))
        fputs(linha, stdout);
    fclose(f);
}

void append_log(const char *path, const char *msg) {
    FILE *f = fopen(path, "a");
    if (!f) return;
    fprintf(f, "%s\n", msg);
    fclose(f);
}

int main(void) {
    Aluno turma[] = {
        {"Ana Souza", 8.7}, {"Bruno Lima", 7.4},
        {"Carla Neves", 9.2}, {"Diego Melo", 6.5}
    };

    salvar_csv("alunos.csv", turma, 4);
    ler_csv("alunos.csv");

    /* Arquivo binário */
    printf("\n=== Arquivo Binário ===\n");
    FILE *fb = fopen("alunos.bin", "wb");
    fwrite(turma, sizeof(Aluno), 4, fb);
    fclose(fb);

    Aluno lidos[4];
    fb = fopen("alunos.bin", "rb");
    int lidos_n = fread(lidos, sizeof(Aluno), 4, fb);
    fclose(fb);
    printf("Lidos do binário: %d registros\n", lidos_n);
    for (int i = 0; i < lidos_n; i++)
        printf("  %s: %.2f\n", lidos[i].nome, lidos[i].media);

    /* Log com append */
    append_log("log.txt", "[INFO] Sistema iniciado");
    append_log("log.txt", "[INFO] Dados carregados");
    ler_csv("log.txt");

    return 0;
}
EOF
gcc arquivos.c -o arquivos && ./arquivos
````

## 📌 14 — Conexão com Banco de Dados (SQLite) <a name='s14'></a>

**Teoria:**  
SQLite é um banco de dados embutido — armazena tudo em um único arquivo `.db`, sem necessidade de servidor. É ideal para aplicativos desktop, mobile e embarcados.

**API básica:**
- `sqlite3_open(path, &db)` — abre/cria o banco  
- `sqlite3_exec(db, sql, callback, 0, &err)` — executa SQL simples  
- `sqlite3_prepare_v2` + `sqlite3_bind_*` — **Prepared Statements** (evita SQL Injection)  
- `sqlite3_step` — executa e avança pelo resultado  
- `sqlite3_column_*` — lê valores da linha atual  
- `sqlite3_finalize` — libera o statement  
- `sqlite3_close(db)` — fecha o banco  

⚠️ **Nunca** concatene strings do usuário diretamente no SQL. Use Prepared Statements.

````python
%%shell
# Verifica se sqlite3-dev está disponível
dpkg -l libsqlite3-dev | grep -q '^ii' && echo 'SQLite OK' || (apt-get install -y -q libsqlite3-dev && echo 'SQLite instalado')
````

````python
%%shell
cat << 'EOF' > db_sqlite.c
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <sqlite3.h>

/* ─── Helpers ─────────────────────────────────────────── */
static void check(int rc, sqlite3 *db, const char *msg) {
    if (rc != SQLITE_OK && rc != SQLITE_DONE && rc != SQLITE_ROW) {
        fprintf(stderr, "Erro %s: %s\n", msg, sqlite3_errmsg(db));
        sqlite3_close(db);
        exit(1);
    }
}

/* ─── CREATE ───────────────────────────────────────────── */
void criar_tabela(sqlite3 *db) {
    const char *sql =
        "CREATE TABLE IF NOT EXISTS alunos ("
        "  id    INTEGER PRIMARY KEY AUTOINCREMENT,"
        "  nome  TEXT NOT NULL,"
        "  curso TEXT,"
        "  media REAL CHECK(media >= 0 AND media <= 10)"
        ");";
    char *err = NULL;
    check(sqlite3_exec(db, sql, NULL, 0, &err), db, "criar_tabela");
    printf("Tabela 'alunos' pronta.\n");
}

/* ─── INSERT com Prepared Statement ───────────────────── */
void inserir(sqlite3 *db, const char *nome, const char *curso, double media) {
    const char *sql = "INSERT INTO alunos(nome, curso, media) VALUES(?, ?, ?);";
    sqlite3_stmt *stmt;

    check(sqlite3_prepare_v2(db, sql, -1, &stmt, NULL), db, "prepare INSERT");
    sqlite3_bind_text(stmt, 1, nome,  -1, SQLITE_STATIC);
    sqlite3_bind_text(stmt, 2, curso, -1, SQLITE_STATIC);
    sqlite3_bind_double(stmt, 3, media);

    check(sqlite3_step(stmt), db, "step INSERT");
    sqlite3_finalize(stmt);
}

/* ─── SELECT ───────────────────────────────────────────── */
void listar(sqlite3 *db, const char *filtro_curso) {
    const char *sql = filtro_curso
        ? "SELECT id, nome, curso, media FROM alunos WHERE curso = ? ORDER BY media DESC;"
        : "SELECT id, nome, curso, media FROM alunos ORDER BY media DESC;";

    sqlite3_stmt *stmt;
    check(sqlite3_prepare_v2(db, sql, -1, &stmt, NULL), db, "prepare SELECT");
    if (filtro_curso) sqlite3_bind_text(stmt, 1, filtro_curso, -1, SQLITE_STATIC);

    printf("\n%-4s %-22s %-18s %s\n", "ID", "Nome", "Curso", "Média");
    printf("%-4s %-22s %-18s %s\n", "----", "----------------------", "------------------", "-----");

    int rows = 0;
    while (sqlite3_step(stmt) == SQLITE_ROW) {
        printf("%-4d %-22s %-18s %.2f\n",
               sqlite3_column_int(stmt, 0),
               sqlite3_column_text(stmt, 1),
               sqlite3_column_text(stmt, 2),
               sqlite3_column_double(stmt, 3));
        rows++;
    }
    printf("%d registro(s) encontrado(s).\n", rows);
    sqlite3_finalize(stmt);
}

/* ─── UPDATE ───────────────────────────────────────────── */
void atualizar_media(sqlite3 *db, int id, double nova_media) {
    const char *sql = "UPDATE alunos SET media = ? WHERE id = ?;";
    sqlite3_stmt *stmt;
    check(sqlite3_prepare_v2(db, sql, -1, &stmt, NULL), db, "prepare UPDATE");
    sqlite3_bind_double(stmt, 1, nova_media);
    sqlite3_bind_int(stmt, 2, id);
    check(sqlite3_step(stmt), db, "step UPDATE");
    sqlite3_finalize(stmt);
    printf("\nAluno id=%d atualizado: média = %.2f\n", id, nova_media);
}

/* ─── DELETE ───────────────────────────────────────────── */
void deletar(sqlite3 *db, int id) {
    const char *sql = "DELETE FROM alunos WHERE id = ?;";
    sqlite3_stmt *stmt;
    check(sqlite3_prepare_v2(db, sql, -1, &stmt, NULL), db, "prepare DELETE");
    sqlite3_bind_int(stmt, 1, id);
    check(sqlite3_step(stmt), db, "step DELETE");
    sqlite3_finalize(stmt);
    printf("Aluno id=%d removido. Linhas afetadas: %d\n",
           id, (int)sqlite3_changes(db));
}

/* ─── AGREGAÇÃO ────────────────────────────────────────── */
void estatisticas(sqlite3 *db) {
    const char *sql =
        "SELECT curso,"
        "       COUNT(*) as qtd,"
        "       AVG(media) as media_turma,"
        "       MAX(media) as maior,"
        "       MIN(media) as menor"
        " FROM alunos GROUP BY curso;";
    sqlite3_stmt *stmt;
    check(sqlite3_prepare_v2(db, sql, -1, &stmt, NULL), db, "prepare GROUP");

    printf("\n=== Estatísticas por Curso ===\n");
    printf("%-18s %4s %8s %6s %6s\n", "Curso", "Qtd", "Média", "Max", "Min");
    while (sqlite3_step(stmt) == SQLITE_ROW) {
        printf("%-18s %4d %8.2f %6.2f %6.2f\n",
               sqlite3_column_text(stmt, 0),
               sqlite3_column_int(stmt, 1),
               sqlite3_column_double(stmt, 2),
               sqlite3_column_double(stmt, 3),
               sqlite3_column_double(stmt, 4));
    }
    sqlite3_finalize(stmt);
}

/* ─── MAIN ─────────────────────────────────────────────── */
int main(void) {
    sqlite3 *db;
    check(sqlite3_open(":memory:", &db), db, "open");  /* :memory: = banco na RAM */
    printf("=== Banco SQLite (in-memory) aberto ===\n\n");

    criar_tabela(db);

    /* INSERTs */
    printf("\n=== Inserindo registros ===\n");
    inserir(db, "Ana Souza",    "Ciência da Computação", 8.7);
    inserir(db, "Bruno Lima",   "Engenharia de Software", 7.4);
    inserir(db, "Carla Neves",  "Ciência da Computação", 9.2);
    inserir(db, "Diego Melo",   "Sistemas de Informação", 6.5);
    inserir(db, "Elisa Prado",  "Engenharia de Software", 8.1);
    inserir(db, "Fábio Costa",  "Ciência da Computação", 5.9);
    printf("6 alunos inseridos.\n");

    /* SELECT todos */
    printf("\n=== Todos os alunos (ordem por média) ===");
    listar(db, NULL);

    /* SELECT filtrado */
    printf("\n=== Filtrado: Ciência da Computação ===");
    listar(db, "Ciência da Computação");

    /* UPDATE */
    atualizar_media(db, 2, 8.0);

    /* DELETE */
    deletar(db, 6);

    /* Resultado final */
    printf("\n=== Estado final ===");
    listar(db, NULL);

    /* Estatísticas */
    estatisticas(db);

    sqlite3_close(db);
    printf("\nBanco fechado. Fim!\n");
    return 0;
}
EOF
gcc db_sqlite.c -lsqlite3 -o db_sqlite && ./db_sqlite
````

## 📌 15 — Boas Práticas e Compilação <a name='s15'></a>

**Flags de compilação recomendadas:**

| Flag | O que faz |
|------|-----------|
| `-Wall` | Ativa a maioria dos avisos |
| `-Wextra` | Avisos adicionais |
| `-std=c17` | Padrão C17 |
| `-g` | Símbolos de debug (gdb) |
| `-O2` | Otimização para produção |
| `-lm` | Linka libm (math.h) |
| `-fsanitize=address` | Detecta acesso inválido à memória |
| `-Werror` | Trata avisos como erros |

**Convenções de nomenclatura:**
- Variáveis e funções: `snake_case` → `calcular_media`  
- Constantes: `MAIUSCULAS` → `MAX_ALUNOS`  
- Tipos (`typedef struct`): `PascalCase` → `Aluno`

````python
%%shell
cat << 'EOF' > boas_praticas.c
/*
 * boas_praticas.c
 * Demonstra: comentários, #define, guard include,
 * assert, math.h, macros seguras e compilação robusta.
 *
 * Compile: gcc -Wall -Wextra -std=c17 boas_praticas.c -lm -o app
 */
#include <stdio.h>
#include <stdlib.h>
#include <math.h>
#include <assert.h>

/* Constantes em MAIÚSCULAS via #define */
#define PI           3.14159265358979
#define MAX_ALUNOS   100
#define VERSAO       "1.0.0"

/* Macro com parênteses — evita surpresas na precedência */
#define QUADRADO(x)  ((x) * (x))
#define MAX(a, b)    ((a) > (b) ? (a) : (b))
#define CLAMP(v,lo,hi) ((v)<(lo)?(lo):(v)>(hi)?(hi):(v))

/* Função com nome descritivo em snake_case */
double area_circulo(double raio) {
    assert(raio >= 0 && "raio não pode ser negativo");
    return PI * QUADRADO(raio);
}

double hipotenusa(double a, double b) {
    return sqrt(QUADRADO(a) + QUADRADO(b));
}

/* Enum para legibilidade */
typedef enum { REPROVADO, APROVADO } Situacao;

Situacao classificar(double media) {
    return (media >= 7.0) ? APROVADO : REPROVADO;
}

int main(void) {
    printf("Versão: %s\n\n", VERSAO);

    /* Macros */
    printf("=== Macros ===\n");
    printf("QUADRADO(7)     = %d\n", QUADRADO(7));
    printf("MAX(3, 8)       = %d\n", MAX(3, 8));
    printf("CLAMP(150,0,100)= %d\n", CLAMP(150, 0, 100));

    /* math.h */
    printf("\n=== math.h ===\n");
    printf("PI              = %.10f\n", PI);
    printf("área círculo r=5= %.4f\n",  area_circulo(5.0));
    printf("hipotenusa(3,4) = %.1f\n",  hipotenusa(3.0, 4.0));
    printf("sqrt(2)         = %.6f\n",  sqrt(2.0));
    printf("pow(2,10)       = %.0f\n",  pow(2.0, 10.0));
    printf("log2(1024)      = %.1f\n",  log2(1024.0));
    printf("ceil(3.2)       = %.0f\n",  ceil(3.2));
    printf("floor(3.9)      = %.0f\n",  floor(3.9));
    printf("fabs(-7.5)      = %.1f\n",  fabs(-7.5));

    /* Enum */
    printf("\n=== Enum ===\n");
    double medias[] = {9.0, 6.5, 7.0, 4.9};
    for (int i = 0; i < 4; i++)
        printf("média %.1f → %s\n", medias[i],
               classificar(medias[i]) == APROVADO ? "Aprovado" : "Reprovado");

    return 0;
}
EOF
gcc -Wall -Wextra -std=c17 boas_praticas.c -lm -o boas_praticas && ./boas_praticas
````

## 🧪 Desafio Final — Mini Sistema de Cadastro

Integra tudo: structs, alocação dinâmica, arquivos e SQLite.

````python
%%shell
cat << 'EOF' > sistema.c
/*
 * Mini Sistema de Cadastro de Alunos
 * Integra: structs, alocação dinâmica, SQLite.
 * Compile: gcc sistema.c -lsqlite3 -o sistema
 */
#include <stdio.h>
#include <stdlib.h>
#include <string.h>
#include <sqlite3.h>

#define DB_PATH  ":memory:"
#define MAX_NOME 80

typedef struct {
    int    id;
    char   nome[MAX_NOME];
    double media;
} Aluno;

/* ─── CRUD ─────────────────────────────────────────────── */
sqlite3 *abrir_banco(void) {
    sqlite3 *db;
    if (sqlite3_open(DB_PATH, &db) != SQLITE_OK) {
        fprintf(stderr, "Erro ao abrir banco: %s\n", sqlite3_errmsg(db));
        exit(1);
    }
    sqlite3_exec(db,
        "CREATE TABLE IF NOT EXISTS alunos("
        "id INTEGER PRIMARY KEY AUTOINCREMENT,"
        "nome TEXT NOT NULL, media REAL);",
        NULL, 0, NULL);
    return db;
}

long inserir_aluno(sqlite3 *db, const char *nome, double media) {
    sqlite3_stmt *s;
    sqlite3_prepare_v2(db,
        "INSERT INTO alunos(nome,media) VALUES(?,?);", -1, &s, NULL);
    sqlite3_bind_text(s, 1, nome, -1, SQLITE_STATIC);
    sqlite3_bind_double(s, 2, media);
    sqlite3_step(s);
    sqlite3_finalize(s);
    return (long)sqlite3_last_insert_rowid(db);
}

/* Retorna array alocado dinamicamente; *out_n recebe o tamanho */
Aluno *buscar_todos(sqlite3 *db, int *out_n) {
    sqlite3_stmt *s;
    sqlite3_prepare_v2(db,
        "SELECT id,nome,media FROM alunos ORDER BY media DESC;",
        -1, &s, NULL);

    int cap = 8, n = 0;
    Aluno *v = malloc(cap * sizeof(Aluno));

    while (sqlite3_step(s) == SQLITE_ROW) {
        if (n == cap) { cap *= 2; v = realloc(v, cap * sizeof(Aluno)); }
        v[n].id = sqlite3_column_int(s, 0);
        strncpy(v[n].nome, (const char*)sqlite3_column_text(s, 1), MAX_NOME-1);
        v[n].media = sqlite3_column_double(s, 2);
        n++;
    }
    sqlite3_finalize(s);
    *out_n = n;
    return v;
}

void imprimir_relatorio(Aluno *v, int n) {
    int aprov = 0;
    double soma = 0;
    printf("\n%s\n", "══════════════════════════════════════════");
    printf(" %-4s %-24s %6s  %s\n", "ID", "Nome", "Média", "Sit.");
    printf("%s\n", "──────────────────────────────────────────");
    for (int i = 0; i < n; i++) {
        int ok = v[i].media >= 7.0;
        printf(" %-4d %-24s %6.2f  %s\n",
               v[i].id, v[i].nome, v[i].media,
               ok ? "✓ Aprovado" : "✗ Reprovado");
        aprov += ok;
        soma  += v[i].media;
    }
    printf("%s\n", "──────────────────────────────────────────");
    printf(" Total: %d  |  Aprovados: %d  |  Média geral: %.2f\n",
           n, aprov, n ? soma/n : 0.0);
    printf("%s\n", "══════════════════════════════════════════");
}

int main(void) {
    sqlite3 *db = abrir_banco();

    /* Dados de exemplo */
    const char *nomes[] = {
        "Ana Souza","Bruno Lima","Carla Neves",
        "Diego Melo","Elisa Prado","Fábio Costa",
        "Gabriela Ramos","Henrique Torres"
    };
    double medias[] = {8.7, 7.4, 9.2, 6.5, 8.1, 5.9, 7.8, 9.5};
    int qtd = 8;

    printf("Inserindo %d alunos...\n", qtd);
    for (int i = 0; i < qtd; i++) {
        long id = inserir_aluno(db, nomes[i], medias[i]);
        printf("  id=%ld  %s\n", id, nomes[i]);
    }

    /* Busca e relatório */
    int n;
    Aluno *turma = buscar_todos(db, &n);
    imprimir_relatorio(turma, n);
    free(turma);

    sqlite3_close(db);
    return 0;
}
EOF
gcc sistema.c -lsqlite3 -o sistema && ./sistema
````
